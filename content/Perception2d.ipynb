# Perception 2D

For determining the type, and possibly the distance of objects, a YOLO26 ('You Only Look Once-26') object-detection model was used. A dataset was created by recording videos using the USB camera included in the MIRTE Master hardware package, from which one in every fifteen frames was extracted for the training dataset.
