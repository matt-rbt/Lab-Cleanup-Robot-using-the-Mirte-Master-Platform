# Perception 2D

For determening the type, and possibly, the distance of objects we use a YOLO26 object classification model. We made our own dataset by recording video using the USB Camera included in the MIRTE Master hardware package. After recording the video we extracted one in every fifteen frames to use for our training dataset. A picture of 4 frames in different environments is shown below

```{figure} https://github.com/matt-rbt/Lab-Cleanup-Robot-using-the-Mirte-Master-Platform/blob/main/content/figures/trainingdataset.png
:label: fig_trainingdataset
:width: 70%
:align: center

Picture of 4 individual frames
```

Training was done manually using Label Studio. Inittially, ~280 frames were captured in 4 different environments. Due to a large amount of blurring in the images, about 100 pictures were deleted. Of the remaining 180 pictures, 24 pictures were imported into Label Studio and labelled per colour: Green, Red, Purple, White, Light grey, Dark grey, and Black. In total 509 bounding boxes were drawn. The relevant results of the training session are shown below. 

```{figure} https://github.com/matt-rbt/Lab-Cleanup-Robot-using-the-Mirte-Master-Platform/blob/main/content/figures/trainingdatasetresults.png
:label: fig_trainingdatasetresults
:width: 70%
:align: center

Picture of tests performed during training
```
